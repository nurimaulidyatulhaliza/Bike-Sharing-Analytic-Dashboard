## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".

- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

**Mengganti nama kolom agar lebih jelas dan mudah dibaca**

```python
day_df.rename(columns={
    'dteday': 'dateday',
    'yr': 'year',
    'mnth': 'month',
    'hum': 'humidity',
    'cnt': 'total_count'
}, inplace=True)

hour_df.rename(columns={
    'dteday': 'dateday',
    'yr': 'year',
    'mnth': 'month',
    'hr': 'hour',
    'hum': 'humidity',
    'cnt': 'total_count'
}, inplace=True)
```

**Menghapus kolom 'instant' yang hanya berisi nomor indeks**

```python
day_df.drop(columns=['instant'], inplace=True)
hour_df.drop(columns=['instant'], inplace=True)
```

**Mengubah tipe data kolom 'dteday' menjadi datetime**

```python
day_df['dateday'] = pd.to_datetime(day_df['dateday'])
hour_df['dateday'] = pd.to_datetime(hour_df['dateday'])
```

**Mengubah kode angka menjadi label keterangan yang mudah dibaca**

```python
season_map = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
yr_map = {0: 2011, 1: 2012}
mnth_map = {1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
            7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'}
weekday_map = {0: 'Sunday', 1: 'Monday', 2: 'Tuesday', 3: 'Wednesday',
               4: 'Thursday', 5: 'Friday', 6: 'Saturday'}
weathersit_map = {
    1: 'Clear / Partly Cloudy',
    2: 'Mist / Cloudy',
    3: 'Light Snow / Rain',
    4: 'Heavy Rain / Snow'
}
workingday_map = {0: 'Holiday / Weekend', 1: 'Working Day'}

for df in [day_df, hour_df]:
    df['season'] = df['season'].map(season_map)
    df['year'] = df['year'].map(yr_map)
    df['month'] = df['month'].map(mnth_map)
    df['weekday'] = df['weekday'].map(weekday_map)
    df['weathersit'] = df['weathersit'].map(weathersit_map)
    df['workingday'] = df['workingday'].map(workingday_map)
```

**Mengembalikan nilai suhu, kelembapan, dan kecepatan angin ke skala asli**

```python
for df in [day_df, hour_df]:
    df['temp'] = df['temp'] * 41
    df['atemp'] = df['atemp'] * 50
    df['humidity'] = df['humidity'] * 100
    df['windspeed'] = df['windspeed'] * 67
```

### Pertanyaan 3:
Berapa persen kenaikan total penyewaan sepeda setiap bulannya dari tahun 2011 ke tahun 2012 untuk mengevaluasi apakah perusahaan perlu menambah jumlah unit sepeda baru?

```python
plt.figure(figsize=(12, 6))

monthly_trend = day_df.groupby(['year', 'month'], observed=False)['total_count'].sum().reset_index()

sns.lineplot(
    data=monthly_trend,
    x='month',
    y='total_count',
    hue='year',
    marker='o',
    palette={2011: '#8c969e', 2012: '#1f77b4'},
    linewidth=3
)

# Menambahkan label persentase kenaikan di atas titik tahun 2012
data_2011 = monthly_trend[monthly_trend['year'] == 2011].reset_index(drop=True)
data_2012 = monthly_trend[monthly_trend['year'] == 2012].reset_index(drop=True)

for i in range(len(data_2012)):
    pct = ((data_2012.loc[i, 'total_count'] - data_2011.loc[i, 'total_count']) / data_2011.loc[i, 'total_count']) * 100
    plt.text(i, data_2012.loc[i, 'total_count'] + 5000, f"+{pct:.1f}%", ha='center', fontsize=9, fontweight='bold', color='#1f77b4')

plt.title('Perbandingan Tren Total Penyewaan Sepeda Bulanan (2011 vs 2012)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Bulan', fontsize=11)
plt.ylabel('Total Penyewaan Sepeda', fontsize=11)
plt.ylim(0, 245000)  # Memberi ruang di atas agar teks persen tidak terpotong
plt.legend(title='Tahun', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**inisght**
Jumlah penyewaan sepeda pada tahun 2012 meningkat 64,88% dibandingkan tahun 2011. Kenaikan terjadi di semua bulan, dengan lonjakan tertinggi pada bulan Maret (157,4%). Penyewaan paling ramai terjadi pada September dengan lebih dari 218 ribu penyewaan. Artinya, minat masyarakat semakin tinggi, sehingga perusahaan perlu menambah stok sepeda agar kebutuhan pengguna tetap terpenuhi.

## Analisis Lanjutan (Opsional)

Analisis Lanjutan (Manual Clustering & Binning)

**1. Pengelompokan Periode Waktu Operasional (Time of Day Grouping)**

```python
# Mengelompokkan jam (0-23) menjadi 5 kategori periode waktu operasional
def categorize_time(hour):
    if 0 <= hour < 6:
        return '1. Dini Hari (00-05)'
    elif 6 <= hour < 10:
        return '2. Jam Sibuk Pagi (06-09)'
    elif 10 <= hour < 16:
        return '3. Siang Hari (10-15)'
    elif 16 <= hour < 20:
        return '4. Jam Sibuk Sore (16-19)'
    else:
        return '5. Malam Hari (20-23)'

hour_df['time_period'] = hour_df['hour'].apply(categorize_time)

time_cluster = hour_df.groupby('time_period', observed=False).agg({
    'casual': 'mean',
    'registered': 'mean',
    'total_count': ['mean', 'sum']
}).reset_index()

time_cluster.columns = ['Periode Waktu', 'Rata-rata Casual', 'Rata-rata Registered', 'Rata-rata Total', 'Total Penyewaan']
display(time_cluster)
```

**insight :**

Jam Sibuk Sore (16.00–19.00) dan Siang Hari (10.00–15.00) menjadi waktu paling padat sehingga membutuhkan kesiapan petugas penuh, sedangkan periode paling sepi pada Dini Hari (00.00–05.00) (rata-rata 24,91 sepeda/jam) ideal dimanfaatkan untuk perawatan dan pengisian ulang stok sepeda.

**2. Pengelompokan Tingkat Permintaan Harian Berdasarkan Suhu (Demand Level Binning)**

```python
# Membagi total penyewaan harian menjadi 3 kategori tingkat keramaian
day_df['demand_level'] = pd.qcut(day_df['total_count'], q=3, labels=['Low Demand', 'Medium Demand', 'High Demand'])

demand_summary = day_df.groupby('demand_level', observed=False).agg({
    'total_count': ['count', 'min', 'max', 'mean'],
    'temp': 'mean',
    'humidity': 'mean'
}).reset_index()

demand_summary.columns = ['Kategori Permintaan', 'Jumlah Hari', 'Min Sewa', 'Max Sewa', 'Rata-rata Sewa', 'Rata-rata Suhu (°C)', 'Rata-rata Kelembapan (%)']
display(demand_summary)
```

**Visualisasi Hasil Pengelompokan Operasional & Permintaan**
